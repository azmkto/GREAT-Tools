## 1. Load and validate

`validate_export()` fails loudly at load time if a column is missing or a
Sentiment/Media value is outside the canonical vocabulary — instead of that turning into a
silent `NaN` forty cells later.

Note the platform normalisation happens inside `prep.prepare_data()` via the library's
`MEDIA_MAP`, which covers `'Media Mainstream'` and `'Thread'`. The old notebook's local
copy did not, and would `KeyError` on `plat_colors` for exports containing them.

````python
DATA = r'C:\Users\ASUS\Documents\Research Project\Data\Prabowo 2026'

df = pd.read_excel(rf'{DATA}\prabowo26_agustus1.xlsx', header=1)
df = validate_export(df)

# Research object name, used in every plot title.
# (The original notebook used input() here, which blocks an unattended run.)
robj = 'Prabowo'

df.shape
````

## 2. Reshape

`prep` returns the three frames the plots consume. `prepare_data` also hands back the
period labels already formatted for the titles.

````python
df1, start_date, end_date = prep.prepare_data(df)

sent_data = prep.sentiment_data(df1)    # Date x Sentiment (wide)
plat_data = prep.platform_data(df1)     # (Date, Media) -> Count
ps_data   = prep.plat_sent_data(df1)    # Media, Sentiment, Count (long)

print(f'{start_date} - {end_date}')
````

## 3. Sanity checks

The same checks both notebooks ran inline, now as functions. They report rather than raise —
`validate_export()` above is the fail-fast version.

````python
checks.frame_info(df1, 'Main Dataframe')
checks.frame_info(sent_data, 'Sentiment Dataframe')
checks.frame_info(plat_data, 'Platform Dataframe')
checks.frame_info(ps_data, 'Platform-Sentiment Dataframe')
````

````python
checks.check_completeness(df1)
checks.check_coverage(df1)
checks.check_composition(df1)
checks.check_labels(df1)
````

## 4. The overview figure

`monthly_overview` is `weekly_overview` plus a sentiment-trend panel across the top;
`interval` sets the day spacing of that panel's x-axis ticks.

Because the class lists and palettes default to the library values, the call can be this

````python
sent_pct_df, platform_totals_df = monthly_overview(
    ps_data=ps_data,
    platform_data=plat_data,
    sentiment_data=sent_data,
    robj=robj,
    start_date=start_date,
    end_date=end_date,
    interval=3,
)
````

## 6. Word clouds

`sentiment_wordclouds()` runs the whole pipeline in one call — clean, TF-IDF, score, draw.

The words shown are **not** raw frequencies. `distinctive_terms()` scores each term by its
mean TF-IDF *inside* a sentiment minus its mean *outside*, so what you see is the vocabulary
that separates positive from negative, not the words both share. It also drops any unigram
already covered by a surviving bigram, so `"upacara peringatan"` appears once as a phrase
instead of `"upacara"` and `"peringatan"` competing separately.

Cleaning is not a parameter — the pipeline always uses `great.text.clean_for_wordcloud()`,
which is the cleaner specialised for this. It strips the research-object name and the
display-noise stopwords that would otherwise dominate the picture; `clean_for_topics()`
deliberately keeps them, which is why the two are separate functions.

By default `News` is excluded (syndicated copy is long and editorially uniform, so it
swamps social chatter) and only positive/negative are drawn — neutral has no distinctive
vocabulary worth plotting.

````python
fig, terms_by_sentiment = sentiment_wordclouds(
    df1,
    sentiments=('positive', 'negative'),
    sent_colors=sent_colors,     # tints each cloud with its sentiment color
    exclude_media=('News',),     # pass () to keep news copy in
    n=200,                       # terms scored per sentiment
    max_words=100,               # terms actually drawn
)
````

### What drove the picture

`sentiment_wordclouds()` also returns the scored terms, so the cloud is inspectable
rather than just decorative.

````python
for sentiment, scored in terms_by_sentiment.items():
    print(f'\n{sentiment.upper()} — top 15 distinctive terms')
    print(scored.head(15).round(4).to_string())
````

### Running it step by step

The same pipeline, unpacked — useful when you want the intermediate frames, e.g. to reuse
one TF-IDF fit across several groupings.

````python
corpus = wordcloud.prepare_corpus(df1)          # dropna, dedupe, drop News, clean
X, terms = wordcloud.tfidf_matrix(corpus['clean_text'])
mask = corpus['Sentiment'].str.lower() == 'positive'
top = wordcloud.distinctive_terms(X, terms, mask, n=200)

print(f'corpus     : {len(corpus):,} rows')
print(f'tfidf      : {X.shape[0]:,} docs x {X.shape[1]:,} terms')
print(f'distinctive: {len(top)} terms for positive')
top.head(10).round(4)
````
